```python
from pyspark.sql.functions import (
    year, month, dayofmonth,
    to_date, to_timestamp, datediff,
    current_date, date_add, date_sub,
    months_between, date_format, col
)

# year          - extracts the year from a date or timestamp
# month         - extracts the month number (1-12)
# dayofmonth    - extracts the day of the month (1-31)
# to_date       - converts a string to a date
# to_timestamp  - converts a string to a timestamp
# datediff      - calculates days between two dates
# current_date  - returns the current date
# date_add      - adds days to a date
# date_sub      - subtracts days from a date
# months_between - calculates the months between two dates
# date_format   - formats a date or timestamp as a string
```

```python
data = [
    (1, "Rahul", "2024-01-15", "2024-02-20"),
    (2, "Priya", "2023-05-10", "2023-06-25"),
    (3, "Amit",  "2022-12-01", "2023-01-15"),
    (4, "Neha",  "2024-03-22", "2024-04-05"),
    (5, "Rohan", "2021-07-10", "2021-08-30"),
]

columns = [
    "id", "name", "start_date", "end_date"
]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
df.select(year("start_date").alias("Year Extract")).show()
df.select(month("start_date").alias("Year Extract")).show()


df.select(datediff('end_date' , 'start_date').alias("Date Diff")).show()
```
